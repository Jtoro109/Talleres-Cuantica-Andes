# Metodo Montecarlo
Genera coordenadas aleatorias $(x, y)$ dentro del cuadrado unitario $[0, 1] \times [0, 1]$.

Condición de corte: Evalúa si el punto cae dentro del cuadrante de círculo ($x^2 + y^2 \le 1$).

Aproximación: La proporción de puntos que cumplen la condición respecto al total de iteraciones aproxima el área $A_q = \frac{\pi}{4}$. Al multiplicar esa proporción por 4, se obtiene el valor estimado de $\pi$.

```python
import random
import math

pi = math.pi
decimales = 10
count_circle, count_total, pi_calculado = 0, 0 ,0

while round(pi, decimales) != round (pi_calculado, decimales):
    x = random.random()
    y = random.random()
    if (x**2 + y**2) <= 1:
        count_circle += 1
    count_total += 1
    pi_calculado = (count_circle/count_total)*4
    
print (f"Numero de decimales alcanzados: {pi_calculado} Cantidad de puntos generados: {count_total} ")
```

# Aguja de Bufon

Genera valores aleatorios para la distancia del centro de la aguja a la línea más cercana $d$ y su ángulo de inclinación $\theta$.

Condición de cruce: Evalúa si la aguja de longitud $l$ cruza una de las líneas separadas por una distancia $t$ verificando que $d \le \frac{l}{2} \sin(\theta)$.

Aproximación: La proporción de agujas que cruzan la línea ($p$) respecto al total de lanzamientos estima la probabilidad teórica. Al despejar la ecuación, se obtiene el valor estimado de $\pi = \frac{2l}{pt}$.

```python
import random 
import math

long_agujas = 1
pi = math.pi
decimales = 15
pi_calculado = 0.0
count_cruces, count_total = 0, 0

while round(pi, decimales) != round (pi_calculado, decimales):
    d = random.uniform(0, long_agujas/2)
    theta = random.uniform(0, pi / 2)
    
    if  d <= ((long_agujas / 2) * math.sin(theta)):
        count_cruces += 1
    
    count_total += 1
    if count_cruces > 0:
        pi_calculado = (2 * long_agujas * count_total) / (long_agujas * count_cruces)

print (f"Numero de decimales alcanzados: {pi_calculado} Cantidad de agujas generadas: {count_total} ")
```
